```julia
using PythonCall

pd = pyimport("pandas")
plt = pyimport("matplotlib.pyplot")
ticker = pyimport("matplotlib.ticker")

df = pd.read_csv("C:\\Users\\Thinkpad\\dataset\\tweets.csv", index_col=nothing, encoding="ISO-8859-1")

df

```

```julia
grouped = df.groupby("Target")
#.apply(x -> x.sample(n=200, random_state=42))
df_sample = pd.concat([group.sample(n=200, random_state=42) for (name, group) in grouped])
df_sample.reset_index()
```

```julia

df_sample["Text"] = map(x -> strip(lowercase(replace(string(x), r"(@\w+|\d+|[^\w\s])"=> ""))), df_sample["Text"])
df_filtered = df_sample.filter(["Target", "Text"], axis=1)
```

```julia
nltk = pyimport("nltk")

nltk.download("stopwords")
nltk.download("punkt")
nltk.download("wordnet")
```

```julia
stopwords = nltk.corpus.stopwords.words("english")
lemmatize = nltk.stem.WordNetLemmatizer()


function word_proccess(text) 

    tokens = nltk.word_tokenize(text)
    filtered_tokens = [lemmatize.lemmatize(token) for token in tokens if !(token in stopwords)]
    return filtered_tokens
    
end

df_filtered["Tokens"] = map(word_proccess, df_sample.Text)
df_filtered
```

```julia
sklearn = pyimport("sklearn")
gensim = pyimport("gensim")
np = pyimport("numpy")
```

```julia
model = gensim.models.Word2Vec(sentences = df_filtered["Tokens"], vector_size=100, window= 5, min_count=1, workers=4)

function text_to_vector(tokens, model)
    vectors = [np.copy(model.wv[token]) for token in tokens if haskey(model.wv, token)]
    return return length(vectors) > 0 ? np.mean(vectors, axis=0) : np.zeros(model.vector_size)
end

sentence_vectors = [text_to_vector(tokens, model) for tokens in df_filtered["Tokens"]]
sentence_vectors_df = np.array(sentence_vectors)
df_filtered["sentence_vector"] = sentence_vectors_df

vector = np.array(df_filtered["sentence_vector"].tolist()).copy()
y = np.array(df_filtered["Target"]).copy()

train_test_split = sklearn.model_selection.train_test_split

X_train, X_test, y_train, y_test = train_test_split(vector, y, test_size=0.4, random_state=42)
```

```julia
svm = sklearn.svm.SVC()
GridSearchCV = sklearn.model_selection.GridSearchCV

param = Dict(
    "C" => [0.1, 1, 10],
    "kernel" => ["linear", "rbf"],
    "gamma" => ["scale", "auto"]
)

params = PythonCall.pydict(param)

gridsearch = GridSearchCV(svm, params, cv=5, scoring="accuracy", error_score = "raise")
gridsearch.fit(X_train, y_train)
```

```julia
println("Best parameters: ", gridsearch.best_params_)
println("Best score: ", gridsearch.best_score_)
y_pred = gridsearch.predict(X_test)
accuracy = sklearn.metrics.accuracy_score(y_test, y_pred)

results = pd.DataFrame(gridsearch.cv_results_)
display(results)
```
